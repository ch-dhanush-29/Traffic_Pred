This notebook explores the arterial corridor telemetry, traffic distributions, weather impacts, and fundamental traffic flow dynamics.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
%matplotlib inline
```

## 1. Load Processed Dataset

```python
df = pd.read_parquet('../data/processed/flowcast_processed.parquet')
print('Dataset shape:', df.shape)
print('Columns:', df.columns.tolist()[:15])
df.head()
```

## 2. Diurnal Traffic Volume Distributions

```python
plt.figure(figsize=(12, 5))
sns.histplot(df['traffic_volume'], bins=50, kde=True, color='#1f77b4')
plt.title('Corridor Traffic Volume Distribution (Vehicles / 30 min)')
plt.xlabel('Traffic Volume')
plt.ylabel('Frequency')
plt.show()
```

## 3. Average Diurnal Profile by Hour

```python
hourly = df.groupby('hour_int')['traffic_volume'].mean()
plt.figure(figsize=(10, 4))
plt.plot(hourly.index, hourly.values, marker='o', color='#d62728', lw=2.5)
plt.title('Corridor 24-Hour Diurnal Volume Curve')
plt.xlabel('Hour of Day')
plt.ylabel('Mean Volume')
plt.grid(True)
plt.show()
```

## 4. Fundamental Diagram: Speed vs. Volume

```python
sample = df.sample(5000, random_state=42)
plt.figure(figsize=(8, 5))
plt.scatter(sample['traffic_volume'], sample['avg_speed'], c=sample['occupancy'], cmap='viridis', alpha=0.5)
plt.colorbar(label='Occupancy (%)')
plt.title('Speed vs Volume Curve (Fundamental Diagram)')
plt.xlabel('Volume')
plt.ylabel('Speed (km/h)')
plt.show()
```

## 5. Weather Conditions vs. Average Speed

```python
weather_stats = df.groupby('weather_condition')[['avg_speed', 'traffic_volume']].mean()
print(weather_stats)
weather_stats['avg_speed'].plot(kind='bar', color='teal', figsize=(8, 4))
plt.title('Impact of Weather on Mean Corridor Speed')
plt.ylabel('Speed (km/h)')
plt.show()
```
